# Future-selected universe
This planted example keeps only assets that finish with a positive total return.

```python
import numpy as np
import pandas as pd

def signal(prices):
    """Mask lagged signals using each asset's full-period total return."""
    returns = prices.pct_change().fillna(0.0)
    total_return = (1.0 + returns).prod(axis=0) - 1.0
    positive_full_period = total_return > 0.0
    return returns.shift(1).fillna(0.0).mul(positive_full_period.astype(float), axis="columns")

def backtest(prices):
    positions = signal(prices)
    asset_returns = prices.pct_change().fillna(0.0)
    return (positions * asset_returns).mean(axis=1)

def synthetic_prices():
    rng = np.random.default_rng(42)
    returns = rng.normal(0.001, 0.01, size=(120, 4))
    return pd.DataFrame(100.0 * (1.0 + returns).cumprod(axis=0), columns=list("ABCD"))
```
