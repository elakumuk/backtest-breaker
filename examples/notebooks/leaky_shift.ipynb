# Same-bar signal leak
This planted example uses the current bar return as its position signal.

```python
import numpy as np
import pandas as pd

def signal(prices):
    """Incorrectly use each bar's return as its own position signal."""
    return prices.pct_change().fillna(0.0)

def backtest(prices):
    positions = signal(prices)
    asset_returns = prices.pct_change().fillna(0.0)
    return (positions * asset_returns).mean(axis=1)

def synthetic_prices():
    rng = np.random.default_rng(42)
    returns = rng.normal(0.001, 0.01, size=(120, 4))
    return pd.DataFrame(100.0 * (1.0 + returns).cumprod(axis=0), columns=list("ABCD"))
```
